```python
import pandas as pd
from surprise import SVD, Dataset, Reader
from surprise.accuracy import mae, rmse
from surprise.model_selection import train_test_split

# Step 1: Load and prepare the MovieLens 100k dataset
data = Dataset.load_builtin("ml-100k")

# Step 2: Preprocess the data by splitting it into 80% training and 20% testing sets
trainset, testset = train_test_split(data, test_size=0.20, random_state=42)

# Step 3: Initialize and train the SVD (Singular Value Decomposition) collaborative filtering model
model = SVD()
print("Training the collaborative filtering model...")
model.fit(trainset)
print("Model training complete!\n")

# Step 4: Predict ratings on the test set and evaluate performance using RMSE and MAE
predictions = model.test(testset)

print("--- Model Performance Metrics ---")
rmse_score = rmse(predictions)  # Calculates Root Mean Squared Error
mae_score = mae(predictions)    # Calculates Mean Absolute Error

# Step 5: Predict a movie rating for a specific unseen user-movie combination
user_id = str(196)
item_id = str(302)
actual_rating = 4

# Generate the predicted rating
pred = model.predict(user_id, item_id, r_ui=actual_rating)

print("\n--- Sample Prediction Output ---")
print(f"User ID: {user_id}")
print(f"Movie ID: {item_id}")
print(f"Predicted Rating: {pred.est:.2f}")
```
